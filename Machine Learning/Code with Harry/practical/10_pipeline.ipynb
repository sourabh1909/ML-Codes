```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('housing.csv')
```

```python
df['income_cat'] = pd.cut(df['median_income'],bins=[0,1.5,3.0,4.5,6.0,np.inf],labels=[1,2,3,4,5])
```

```python
from sklearn.model_selection import StratifiedShuffleSplit
split = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
for train_index, test_index in split.split(df, df['income_cat']):
    strat_train_set = df.loc[train_index]
    strat_test_set = df.loc[test_index]
```

```python
for sett in (strat_train_set, strat_test_set):
    sett.drop('income_cat', axis=1, inplace=True)
```

```python
df = strat_train_set.copy()
```

- Pipeline class is imported from sklearn

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
```

```python
# this pipeline impute value and automatically std,

mypipeline = Pipeline([
    ('impute',SimpleImputer(strategy='median')),
     ('standardize',StandardScaler())
])
```

- How to works  
1. the pipeline takes a list of steps as (name,tranformer) pairs,
2. Names must be unique and should not contain double underscore __ .
3. All intermediate steps must be transformer(i.e must implement fit_transform)
4. The final step can be either a transformer or predictor.

```python
mypipeline.fit_transform(housing)
```
