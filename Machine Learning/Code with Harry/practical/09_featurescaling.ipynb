```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('housing.csv')
```

```python
df['income_cat'] = pd.cut(df['median_income'],bins=[0,1.5,3.0,4.5,6.0,np.inf],labels=[1,2,3,4,5])
```

```python
from sklearn.model_selection import StratifiedShuffleSplit
split = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
for train_index, test_index in split.split(df, df['income_cat']):
    strat_train_set = df.loc[train_index]
    strat_test_set = df.loc[test_index]
```

```python
for sett in (strat_train_set, strat_test_set):
    sett.drop('income_cat', axis=1, inplace=True)
```

```python
df = strat_train_set.copy()
```

```python
df
```

```python
strat_train_set.drop('median_house_value', axis=1)
strat_labels = strat_train_set['median_house_value'].copy()
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='median')
housing_num = df.select_dtypes(include=[np.number])
imputer.fit(housing_num)
```

```python
X=imputer.transform(housing_num)
```

```python
housing = pd.DataFrame(X, columns=housing_num.columns, index=housing_num.index)
```

```python
housing['ocean_proximity'] = df['ocean_proximity']
```

```python
housing
```

```python
set(housing['ocean_proximity'])
```

```python
# ordinary encoding
from sklearn.preprocessing import OrdinalEncoder
```

```python
ordinal_encoder = OrdinalEncoder()
housing_cat = ordinal_encoder.fit_transform(housing)
housing_cat = pd.DataFrame(housing_cat, columns=housing.columns, index=housing_num.index)
housing_cat
```

```python
# one hot encoder
# fetching categorical data
from sklearn.preprocessing import OneHotEncoder

one_hot_encoder = OneHotEncoder()

housing_cat_1hot = one_hot_encoder.fit_transform(housing[['ocean_proximity']])

feature = one_hot_encoder.categories_[0]   # only category names

housing_cat = pd.DataFrame(
    housing_cat_1hot.toarray(),
    columns=feature,
    index=housing.index
)

print(housing_cat.head())
```

```python
housing_cat
```

```python
df = df.merge(housing_cat, left_index=True, right_index=True)
```

```python
df.drop('ocean_proximity',axis=1,inplace=True)
```

```python
# Method one: Min-Max Scaler ( normalization )
from sklearn.preprocessing import MinMaxScaler

min_max_scaler = MinMaxScaler(feature_range=(-1,1)) # instance of min_max_scalar
df_scale = min_max_scaler.fit_transform(df)
df_scale= pd.DataFrame(df_scale,columns=df.columns,index=df.index)
```

```python
# here we adjust all scale
# but min_max scaling is sensitive to outliers

# therefore we use Z-Score-scaling(standardization)

from sklearn.preprocessing import StandardScaler
scalar = StandardScaler()
df_scale = scalar.fit_transform(df)
df_scale= pd.DataFrame(df_scale,columns=df.columns,index=df.index)
```
