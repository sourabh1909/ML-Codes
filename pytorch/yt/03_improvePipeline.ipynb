```python
import torch
import torch.nn as nn

class Model(nn.Module):
    
    def __init__(self,num_feature):
        super().__init__()  # parent class (nn.Module) ke constructure ko invoke kar rahe hai
        # layer
        self.linear = nn.Linear(num_feature,1) # num_feature is a input & output as single neuron
        self.sigmoid = nn.Sigmoid()
       
    # forword pass 
    def forward(self,feature):
        out = self.linear(feature)  # z = w*x+b
        out = self.sigmoid(out)     # a = sigmoid(z)
        return out
```

```python
# create dataset
features = torch.rand(10,5)

# create model
model = Model(features.shape[1])

# call model for forward pass
model(features)
# yethe ek __call__ function override zale ahe
# tyamule model(features) as a function call kela tar forward pass execute hote
```

```python
# show model weights
print(model.linear.weight)
print(model.linear.bias)
```
