```python
df = pd.read_csv('breast-cancer.csv')
```

```python
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,1:],df.iloc[:,0],test_size=0.2)
```

```python
scalar = StandardScaler()
X_train = scalar.fit_transform(X_train)
X_test = scalar.transform(X_test)
```

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

Numpy array to pytorch tensors

```python
import torch
```

```python
X_train_tensor = torch.from_numpy(X_train)
y_train_tensor = torch.from_numpy(y_train)
X_test_tensor = torch.from_numpy(X_test)
y_test_tensor = torch.from_numpy(y_test)
```

```python
X_train_tensor
```

Defining model

```python
class mySimpleNN():
    
    def __init__(self,X):
        self.weights = torch.rand(X.shape[1],1,dtype=torch.float64,requires_grad=True)
        self.bias = torch.zeros(1,dtype=torch.float64,requires_grad=True)
        
    # defination of forword pass
    def forward(self,X):
        z = torch.mm(X,self.weights) + self.bias
        y_pred = torch.sigmoid(z)
        return y_pred
    
    # binary cross entropy loss function
    def binary_cross_entropy(self,y_pred,y_true):
        # clmap prediction to avoid log(0)
        epsilon = 1e-7
        y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)
        loss= -(y_train_tensor * torch.log(y_pred) + (1 - y_train_tensor) * torch.log(1 - y_pred)).mean()
        return loss
```

```python
learning_rate = 0.1
epochs= 10000
```

```python
model = mySimpleNN(X_train_tensor)

# define loop
for epoch in range(epochs):
# take forword pass
    y_pred = model.forward(X_train_tensor)
# loss
    loss = model.binary_cross_entropy(y_pred,y_train_tensor)
# backword pass
    loss.backward()
    
# update parameter
    with torch.no_grad(): 
        model.weights.data -= learning_rate * model.weights.grad.data
        model.bias.data -= learning_rate * model.bias.grad.data

    model.weights.grad.data.zero_()
    model.bias.grad.data.zero_()

# print loss in each epoch
    print(f"epoch {epoch+1} loss: {loss.item()}")
```

```python
# evaluation
with torch.no_grad():
    y_pred_test = model.forward(X_test_tensor)
    y_pred_test = (y_pred_test > 0.5).float()
    accuracy = (y_pred_test.squeeze() == y_test_tensor).float().mean()
    print(f"Test Accuracy: {accuracy.item() * 100:.2f}%")
```
