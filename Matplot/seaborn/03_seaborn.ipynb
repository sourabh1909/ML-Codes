##### Box plot

- in short based on quartiles
- A boxplot is a standard way of displaying the distribution of data based on a five number summary ("minimum",first quartile[Q1], median,third quartile[Q3],"maximum")
- it can tell you about outliers and what their values are. Boxplot can also tell you if your date is symmetrical how tightly your data is grouped and if and how your data is skewed
- minimum = [Q1] - 1.5*IQR
- maximum = [Q3] + 1.5*IQR

```python
# box plot
sns.boxplot(data=tips,x='sex',y='total_bill')
# sns.boxplot(data=tips,x='day',y='total_bill')
sns.boxplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
# catplot -> figure level
sns.catplot(data=tips,x='day',y='total_bill',kind='boxen')
sns.catplot(data=tips,x='day',y='total_bill',kind='box')
```

- the black which is present at middle of the column represents, when there are multiple observations in each category, it also uses botstraping to compute a confidnce interval around the estimate, which is plotted using error bars (repete x axis) 

```python
sns.barplot(data=tips,x='sex',y='total_bill',
            hue='smoker')
# to remove that erroe bar the prameter is used errorbar
```

In the simplest invocation,both(regplot + lmplot) function draw a scatterplot of two variable, x and y,and then fit th regression model y ~ x and plot the resulting line and a 95% confidence interval for that regression

```python
sns.lmplot(data=tips,x='total_bill',y='tip',hue='sex')

# in regplot we cant apply hue parameter 
# but in lmplot we can apply hue parameter
```

```python
# resideplot
# it us based on regplot 
# the dotted line is regression line and it shows distance
# prediction ka error
sns.residplot(data=tips,x='total_bill',y='tip')
```

```python
# multi-grid plot
# facet-grid

# figure level -> relplot -> displot -> catplot-> lmplot
# in tino mai facet(row and col) ka option hota hai
sns.catplot(data=tips,x='sex',y='total_bill',kind='violin',hue='sex',col='day',row='time')
```

```python
# facetgrid
# more customizes

g =sns.FacetGrid(data=tips,col='day',row='time')
g.map(sns.boxplot,'sex','total_bill' )
```

```python
g =sns.FacetGrid(data=tips,col='day',row='time',hue='smoker')
g.map(sns.scatterplot,'tip','total_bill' )
g.add_legend()
```

```python
# pair-plot
# detect numerical columns in your data
# add makes pair plot for each pair with each other
# pair plot it puts like scatter for each pair and if same name is occur then it will put histogram
# ()
iris
sns.pairplot(iris)
sns.pairplot(tips)
```

```python
sns.pairplot(iris,hue='species')
```

```python
# pairplot is a chold of pairgrid 
# in pairgrid we get more options for customization

g = sns.PairGrid(data=iris,hue='species')
g.map(sns.scatterplot)
```

```python
g = sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
# we can plot any type of plot at diagonal and non-diagonal side
```

```python
g = sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

```python
# select specific numeric columns 
g = sns.PairGrid(data=iris,hue='species',vars=['sepal_width','petal_width'])
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

```python
# Joint plot
sns.jointplot(data=tips,x='total_bill',y='tip',kind='kde',hue='sex')
sns.jointplot(data=tips,x='total_bill',y='tip',kind='reg')
# ['scatter', 'hist', 'hex', 'kde', 'reg', 'resid'] these options are available for graphs
```

```python
g = sns.JointGrid(data=tips)
g.plot(sns.scatterplot,sns.histplot)
```

```python
g = sns.JointGrid(data=tips,x='total_bill',y='tip')
g.plot(sns.scatterplot,sns.histplot)
```
