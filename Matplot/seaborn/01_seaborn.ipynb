```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
# we have to connect internet to see this dataset

tips = sns.load_dataset('tips')
tips
```

```python
# scatter plot -> axes level

sns.scatterplot(data=tips,x='total_bill',y='tip')
# it directly provides name to the y and x axis

# there are two types of plot 
# - figure and axis

# now this plot is figure level
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter')
```

```python
# the real power of seaborn
# sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex')
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time')

# all parameter are applied on figure and axes
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',size='size',style='time',hue='sex')
```

```python
# continuty dhikhani hai to line plot use karte hai
# or jub hume wo nahi chahie to hum scattter plot use karte hai

gap = px.data.gapminder()
gap
```

```python
temp_df = gap[gap['country'] == 'India']
```

```python
# axes level data

sns.lineplot(data=temp_df,x='year',y='lifeExp')

# figure level data
sns.relplot(data=temp_df,x='year',y='lifeExp',kind='line')
```

```python
temp_df = gap[gap['country'].isin(['India','Brazil','Germany','China'])]
```

```python
sns.lineplot(data=temp_df,x='year',y='lifeExp',hue='country') # axes
sns.relplot(data=temp_df,x='year',y='lifeExp',hue='country',kind='line') # figure

# in axes plot the legend is in plot
# where as in figure plot the legend is out of this plot
```

```python
# to check where that country lies
sns.relplot(data=temp_df,x='year',y='lifeExp',hue='country',kind='line',style='continent',size='continent') # figure
```

```python
# facet plot
# by using this we can select any perticular category and plot multiple plots
### this is works with figure level functions
# -> it will work with relplot() but not works with axes level function


sns.relplot(data=tips,x='total_bill',y='tip',col='smoker',row='sex')
# sns.relplot(data=tips,x='total_bill',y='tip',row='sex')
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',col='sex',row='day')
```

```python
# col_wrap
sns.relplot(data=gap,x='lifeExp',y='gdpPercap',kind='scatter',col='year',col_wrap=3)
```

2. Distrubution Plots

- data ka distrubution samaj mai help karte hai distrubution plot
- used for univariate analysis
- used to find out the distrubution
- range of the observation
- central Tendency
- is the data bimodel ? (multiple pickes?)
- Are there outliers?

plots under distrubution plot

- histplot
- kdeplot
- rugplot

```python
# figure level function -- ( dis plot )
# axes level function -- ( histplot , kdeplot , rugplot )
```

```python
tips
```

```python
sns.histplot(data=tips,x='total_bill')
```

```python
# figure level

sns.displot(data=tips,x='total_bill',kind='hist')
```

```python
sns.histplot(data=tips,x='total_bill',bins=10)
```

```python
# hist graph can be used to plot graph on categorical data

# count plot 

# element parameter give outline for your graph
sns.histplot(data=tips,x='tip',hue='sex',element='step')

tit = sns.load_dataset('titanic')
```

```python
tit = pd.DataFrame(tit)
tit
```

```python
tit[tit['survived'] == 1][['survived']].count()
```

```python
sns.histplot(data=tit,x= 'age',element='step',hue='sex')
```

```python
# facet using row and col
# will not work on histplot function
# because it axes lavel function
sns.displot(data=tips,x='tip',col='sex',element='step',kind='hist')
```

```python
# kedplot
# kernael denstity estimation
# rather than using discrete bins, a KDE plot smooths the observation with a gaussian kernel , produsing a coutinuous density estimaete

# insimple words it is also represents data distrubution
# in curve
tips
```

```python
sns.kdeplot(data=tips,x='total_bill',hue='sex',fill=True)
# it is more useful than hist plot
# it uses gaussian prob function
```

```python
sns.displot(data=tips,x='total_bill',kind='kde')
```

```python
# rugplot
# not useful but it helps other graph

sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')

# the number line which shown in this called as rugplot
```

```python
# we can also make bivariate histgram
# bivariate histogram bins the data within rectangle that tile the plot
# 
sns.histplot(data=tips,x='total_bill',y='tip',color='blue')
sns.displot(data=tips,x='total_bill',y='tip',color='blue',kind='hist')
```

```python
# bivariate kde plot
# a bivariate KDE plot smooths the (x,y) observation with a 2d gaussian
sns.kdeplot(data=tips,x='total_bill',y='tip')
```

2D matrix plot
- heatmap
- linewidth

```python
# heatmap
# plot rectangle data as a color- ancoded matrix
gap = pd.DataFrame(gap)
```

```python
gap['country'].unique().size
gap['country'].nunique()
```

```python
temp_df = gap.pivot(index='country',columns='year',values='lifeExp')
```

```python
plt.figure(figsize=(15,10))
sns.heatmap(temp_df)
```

```python
temp_df = gap[gap['continent'] == 'Europe'].pivot(index='country',columns='year',values='lifeExp')
```

```python
plt.figure(figsize=(15,20))
sns.heatmap(temp_df,annot=True)
```

```python
plt.figure(figsize=(15,20))
sns.heatmap(temp_df,annot=True,linewidths=0.5)

sns.heatmap(temp_df,annot=True,linewidths=0.5,cmap='jet')
sns.heatmap(temp_df,annot=True,linewidths=0.5,cmap='viridis')
```
