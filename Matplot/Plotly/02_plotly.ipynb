```python
# 3d surface plot
# can not be crated using plotly express
# fro 3d surface plot we are going to use plotly group objects

x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx , yy = np.meshgrid(x,y)

z = xx**2 + yy**2 
z = np.sin(xx) + np.cos(yy)
z = np.sqrt(1-xx**2-yy**2)

trace = go.Surface(x=x,y=y,z=z)

data = [trace]
layout  = go.Layout(title = "3d surface plot")

fig = go.Figure(data,layout)
fig.show()
```

```python
# contour plot 
# gives view from top of the graph

# 3d surface plot
# can not be crated using plotly express
# fro 3d surface plot we are going to use plotly group objects

x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx , yy = np.meshgrid(x,y)

z = xx**2 + yy**2 
# z = np.sin(xx) + np.cos(yy)
# z = np.sqrt(1-xx**2-yy**2)

trace = go.Contour(x=x,y=y,z=z)

data = [trace]
layout  = go.Layout(title = "3d surface plot")

fig = go.Figure(data,layout)
fig.show()
```

```python
# for subplots
from plotly.subplots import make_subplots
```

```python
fig = make_subplots(rows=1,cols=2)
```

```python
# make all begining letters of plot type must be capital
# if it is not then it will throw error as module object is not collab

fig.add_trace(
    go.Scatter(x=[1,3,5],y=[2,10,1]),
    row=1,
    col=1
)
fig.add_trace(
    go.Histogram(x=[1,5,8,9,6,2,74,100,85,22,56,3,4]),
    row=1,
    col=2
)

fig.update_layout(title='subplot demo')
fig.show()
```
